# Map and Filter

```python
# 1. Add 'age_group' column based on 'age'

def age_group(age):
    if age < 30:
        return 'Youth'
    elif 30 <= age < 50:
        return 'Middle-Aged'
    else:
        return 'senior'

df['age_group'] = df['age'].apply(age_group)

```

```python
df[['age_group','age']].head()
```

```python
# 2. Add 'years_at_company'column based on 'length_of_service'

df['years_at_company'] = df['length_of_service']
```

```python
# 3. Add 'training_score_category' column based on 'avg_training_score'
def training_score_category(score):
    if score < 50:
        return 'low'
    elif score <= score < 75:
        return 'Medium'
    else:
        return 'High'

df['training_score_category'] = df['avg_training_score'].apply(training_score_category)
df.head()
```

```python
# 4. Add 'performance column based on 'previous_year_rating'
def performance_rating(rating):
    if rating >=4.0:
        return 'Excellent'
    elif 3.0 <= rating < 4.0:
        return 'Good'
    else:
        return 'Needs Improvement'

df['performance'] =df['previous_year_rating'].apply(performance_rating)
df.head()
```

```python
# 5. Add 'awards_received column based on 'awards_won'
df['awards_received'] = df['awards_won'].apply(lambda x: True if x > 0 else False)
print(df.head())
```

# New Feature Engineering

```python
# Assuming df is already loaded and missing values have been filled

# 1. age_and_servie: Adds the age and length of service to create a combined feature
df['age_and_service'] = df['age'] + df['length_of_service'] #Combined age and serviec to reflect overall experience

# 2. training_efficiency: Measures how efficient an employee is in terms of training
df['training_efficiency'] = df['avg_training_score']/df['no_of_trainings'] # how well the employees perform

#3. service_to_training_ratio: ratio of servie yeard to the number of trainings attended
df['service_to_training_ratio'] = df['length_of_service']/df['no_of_trainings'] # provides insights into training intensity

# 4. age_servie_interaction: Interaction term between age and years of experience
df['age_servie_interaction'] = df['age'] * df['length_of_service'] # This can uncover patterns of employees who have been long

# 5. experience_in_years: Calculates the total experience in years based on the length of service in months
df['experience_in_years'] = df['length_of_service'] / 12

# 6. education_level: Binary feature indcating if an employee had a Master's or PhD
df['education_level'] = df['education'].apply(lambda x: 1 if x in ['Masters', 'PhD'] else 0) 

# 7.re

# 8. is_young_employee: Binary Column marking emplouees who are under 30 years old
df['is_young_employee'] = df['age'].apply(lambda x: 1 if x <30 else 0) # flags employees under 30 years of age

# 9. is_experienced_employee: Binary column marking employees with more than 5 years of service
df['is_experienced_employee'] = df['length_of_service'].apply(lambda x: 1 if x > 5 else 0) #flags employee withmore than 5

#  10. employee_tenure_in_months: Converts length of service into months to reflect tenure in more granular detail
df['employee_tenure_in_months']  = df['length_of_service'] * 12   # converts years to months, showing exact months of employment

# 11. age_times_performance: Combines age and performance to find any relationship between age and performance ratings
df['age_times_performancce'] = df['age'] * df['previous_year_rating'] #multiplies age by performance to explore if age correlate

# 12. service_category: categorizes employee into short-term, medium-term and long-term based on yeard of service
def service_category(years):
    if years < 2:
        return 'Short Term'
    elif 2<= years <= 5:
        return 'Medium Term'
    else:
        return 'Long Term'

df['service_category'] = df['length_of_service'].apply(service_category) # Categories employeed into service periods

# 13. age_category_by_service: Creates a category combining age and years of service for segmentation
def age_service_category(row):
    if row['age'] < 30 and row['length_of_service'] < 2:
        return 'Young Short-term'
    elif row['age'] >= 30 and row['length_of_service'] < 5:
        return 'Mid-age Medium-term'
    elif row['age'] >= 40 and row['length_of_service'] > 5:
        return 'Experienced Long-term'
    else:
        return 'Other'

df['age_category_by_service'] = df.apply(age_service_category, axis=1) # Combines age and service Length

#Show the updated DataFrame
print(df.head())
```
