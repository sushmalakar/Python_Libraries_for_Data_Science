```python
# Sort by multiple columns 
sorted_data_multiple = df.sort_values(by=['SepalLengthCm','SepalWidthCm'], ascending=[False, True])
print("\nSorted data by sepal_length (descending) and sepal_width (ascending):")
print(sorted_data_multiple.head())
```

```python
# Filtering the dataset
# Select rows where petal_length > 1.5

filtered_data = df[df['PetalLengthCm'] > 1.5]
print("\nFiltered data (PetalLengthCm  > 1.5):")
print(filtered_data.head())
```

```python
# Grouping by species and calculating means

grouped_data = df.groupby('Species').mean()
print("\nMean values grouped by species:")
print(grouped_data)
```

```python
# Adding a new column for sepal_area

df['SepalArea'] = df['SepalLengthCm'] * df['SepalWidthCm']
print("\nDataset with SepalArea column added")
print(df.head())
```

```python
# Example: Function to classify sepal length into categories

def classify_sepal_length(length):
    if length < 5.5:
        return 'Short'
    elif length < 6.5:
        return 'Medium'
    else:
        return 'Long'

# Apply function to create a new column 'SepelLengthCategory'

df['SepalLengthCategory'] = df['SepalLengthCm'].apply(classify_sepal_length)
print("\nDataset with SepalLengthCategory column added:")
print(df.head())
```

```python
# Pivot table: mean petal_width by species and sepal_length category

pivot_table = pd.pivot_table(df, values = 'PetalWidthCm', index = 'Species', columns = 'SepalLengthCategory')
print("\nPivot table of mean petal_width by species and SepalLengthCategory:")
print(pivot_table)
```

```python
# Example: Encode species column using one-hot encoding

endoded_data = pd.get_dummies(df, columns=['Species'])
print("\nDataSet with species column added")
print(endoded_data.head())
```

```python
# Example: Reshape data using melt

melted_data = pd.melt(df, id_vars=['Species'], value_vars=['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm'])
print("\nMelted dataset:")
print(melted_data.head())
```

```python
# Visualization (using matplotlib and seaborn)
import seaborn as sns
```

```python
#Scatter plot of sepal_length vs sepal_width
plt.figure(figsize=(8,6))
sns.scatterplot(x='SepalLengthCm', y='SepalWidthCm', data=df, hue='Species', palette='Set2', s=100)
plt.title('Sepal Length vs Width')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
# Box plot of petal_length by species
plt.figure(figsize=(8,6))
sns.boxplot(x='Species',y='PetalLengthCm',data=df,palette='Set2')
plt.title('Petal Length vs Species')
plt.xlabel('Species')
plt.ylabel('Petal Length')
plt.show()
```
