```python
import numpy as np
from matplotlib import pyplot as plt
# import matplotlib.pyplot as plt
```

```python
x = np.arange(1,11)
```

```python
print(x)
```

```python
y = 2*x
```

```python
print(y)
```

```python
plt.plot(x,y)
# plt.show()
```

```python
## Adding Titles and labels
```

```python
plt.plot(x,y)
plt.title("Line Plot of x and y variables")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.show()
```

## changing Line Aesthetics

```python
plt.plot(x,y,color="g",linestyle="*", linewidth=3)
plt.title("Line Plot of x and y variables")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.show()
```
