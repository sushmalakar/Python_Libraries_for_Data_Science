## Making Histogram

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
data=[1,3,3,3,3,9,9,4,4,8,8,8,6,7] # creating a data
```

```python
plt.hist(data, color="b") # plotting a histogram with hist function
plt.show()
```

## Adding the bins

```python
plt.figure(figsize=(9,2))
plt.hist(data, color="g", bins=3) # plotting a histogram with hist function
plt.show()
```

## Working with a real world data

```python
import pandas as pd
```

```python
iris_df = pd.read_csv('Data_Sets/Iris.csv')
```

```python
iris_df.head()
```

```python

plt.subplot(2,4,1)
plt.hist(iris_df['SepalLengthCm'], bins=30, color="g")
plt.title("Plotting Sepallength")
plt.xlabel('x')
plt.ylabel('y')

plt.subplot(2,4,2)
plt.hist(iris_df['SepalWidthCm'], bins=30, color="r")
plt.title("Plotting Sepal Width")
plt.xlabel('x')
plt.ylabel('y')

plt.subplot(2,4,3)
plt.hist(iris_df['SepalWidthCm'], bins=30, color="r")
plt.title("Plotting Sepal Width")
plt.xlabel('x')
plt.ylabel('y')
```
