```python
import numpy as np
from matplotlib import pyplot as plt
# import matplotlib.pyplot as plt
```

```python
x = np.arange(1,11)
```

```python
print(x)
```

```python
y = 2*x
```

```python
print(y)
```

```python
plt.plot(x,y)
# plt.show()
```

```python
## Adding Titles and labels
```

```python
plt.plot(x,y)
plt.title("Line Plot of x and y variables")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.show()
```

## changing Line Aesthetics

```python
plt.plot(x,y,color="g",linestyle=":", linewidth=3)
plt.title("Line Plot of x and y variables")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.show()
```

## Adding two lines in the same plot

```python
y1 = 2*x
y2 = 3*x
```

```python
plt.plot(x,y1,color="g", linewidth=3)
plt.plot(x,y2,color="blue", linestyle=":",linewidth=5)

plt.grid(True)
```

## Adding subpots horizontally

```python
plt.subplot(1,6,1)
plt.plot(x,y1,color="g", linewidth=3)

plt.subplot(1,6,2)
plt.plot(x,y1,color="r", linewidth=3)

plt.subplot(1,6,3)
plt.plot(x,y1,color="orange", linewidth=3)

plt.subplot(1,6,4)
plt.plot(x,y1,color="blue", linewidth=3)

plt.subplot(1,6,5)
plt.plot(x,y1,color="black", linewidth=3)

plt.subplot(1,6,6)
plt.plot(x,y1,color="pink", linewidth=3)

# plt.subplot(2,6,1)
# plt.plot(x,y1,color="g", linewidth=3)


plt.show()
```

## Adding subplots vertically

```python
plt.subplot(2,1,1)
plt.plot(x,y1,color="g",linewidth=3)

plt.subplot(2,1,2)
plt.plot(x,y1,color="lavender",linewidth=3)
plt.show()
```

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("Data_Sets/number.csv", sep=",")
```

```python
df
```

```python
def compute_p(row):
    if row['x'] > 1 and row['y'] > 1 and row['z'] > 1:
        return 3 * row['x'] + 2 * np.exp(row['y']) + 5 * row['z']
    else:
        return 0 # if the condition isn't met, set p to 0

#Compute p and add it as a new column
df['p'] = df.apply(compute_p, axis=1)
```

```python
df.head(60)
```

```python
# Plot p vs x

plt.figure(figsize=(6,6))
plt.plot(df['x'], df['p'], marker='*', linestyle='-', color='b')
plt.title('Line plot of p vs x')
plt.xlabel('x')
plt.ylabel('p')
plt.grid(True)
plt.show()

#Plot p vs y
plt.figure(figsize=(10,6))
plt.plot(df['y'], df['p'], marker='*', linestyle='-', color='r')
plt.title('Line plot of p vs y')
plt.xlabel('y')
plt.ylabel('p')
plt.grid(True)
plt.show()

# plor p vs z
plt.figure(figsize=(5,5))
plt.plot(df['z'], df['p'], marker='o', linestyle='-', color='g')
plt.title('Line plot of p vs z')
plt.xlabel('x')
plt.ylabel('p')
plt.grid(True)
plt.show()
```
