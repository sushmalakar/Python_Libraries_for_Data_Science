## Bar Plot

```python
import matplotlib.pyplot as plt
x = ["python", "C", "C++", "Java"]
y = [85, 70 ,60, 82]
z = [20, 30, 40, 50]
plt.xlabel("Language")
plt.ylabel("Number")
plt.title("Popularity", fontsize=20)

#c =["r","y", "b", "m"]
#plt.bar(x,y,width=0.5, color=c, align="edge", edgecolor="black", linewidth=3, linestyle=":", alpha=0.5, label="popularity")
plt.bar(x,y,width=0.5, color="y", align="edge", edgecolor="black", alpha=0.5, label="popularity")
plt.bar(x,z,width=0.5, color="r", align="edge", edgecolor="black", label="popularity")

# align parameter => edge and center 
# alpha value => 0 to 1
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = ["python", "C", "C++", "Java"]
y = [85, 70 ,60, 82]
z = [20, 30, 40, 50]

width = 0.2
p = np.arange(len(x))
p1 = [j+width for j in p]


plt.xlabel("Language")
plt.ylabel("Number")
plt.title("Popularity", fontsize=20)

plt.bar(p,y,width, color="y", align="edge", edgecolor="black", alpha=0.5, label="popularity")
plt.bar(p1,z,width, color="r", align="edge", edgecolor="black", label="popularity")

plt.xticks(p+width,x, rotation=45)

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = ["python", "C", "C++", "Java"]
y = [85, 70 ,60, 82]
z = [20, 30, 40, 50]
plt.xlabel("Language")
plt.ylabel("Number")
plt.title("Popularity", fontsize=20)

plt.barh(x,y,color="y", align="edge", edgecolor="black", alpha=0.5, label="popularity")
plt.barh(x,z,color="r", align="edge", edgecolor="black", label="popularity")

plt.legend()
plt.show()
```

## Pie chart

```python
import matplotlib.pyplot as plt
x = [10,20,30,40]
y = ["C","C++", "java", "python"]
ex = [0.4, 0.0, 0.0, 0.0]
c = ["r", "b", "y", "g"]
plt.pie(x, labels=y, explode = ex, colors = c, autopct="%0.1f%%", shadow=True, radius = 1.7, labeldistance=0.8)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10,20,30,40]
y = ["C","C++", "java", "python"]
ex = [0.4, 0.0, 0.0, 0.0]
c = ["r", "b", "y", "g"]
plt.pie(x, labels=y, explode = ex, colors = c, autopct="%0.1f%%", shadow=True, radius = 1.7, labeldistance=1.1, 
        textprops={"fontsize":20}, counterclock=False, wedgeprops={"linewidth":4, "edgecolor":"m"}, center=(3,6))
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10,20,30,40]
y = ["C","C++", "java", "python"]
ex = [0.4, 0.0, 0.0, 0.0]
c = ["r", "b", "y", "g"]
plt.pie(x, labels=y, explode = ex, colors = c, autopct="%0.1f%%", shadow=True, labeldistance=1.1, 
        textprops={"fontsize":10}, counterclock=False, wedgeprops={"linewidth":4,}, rotatelabels=True)
plt.title("Pie Chart")
plt.legend(loc=1)
plt.show()
```

## Doughnut piechart

```python
import matplotlib.pyplot as plt
x = [10,20,30,40]
x1 =[40,30,20,10]
y = ["C","C++", "java", "python"]

c= ["r", "b", "g", "y"]

plt.pie(x, labels=y, radius=1.5, colors = c)
plt.pie(x1, radius=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10,20,30,40]
x1 =[40,30,20,10]
y = ["C","C++", "java", "python"]

c= ["r", "b", "g", "y"]

plt.pie(x, labels=y, radius=1.5, colors = c)
plt.pie([1], radius=0.8, colors="w")
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [10,20,30,40]
x1 =[40,30,20,10]
y = ["C","C++", "java", "python"]

c= ["r", "b", "g", "y"]

plt.pie(x, labels=y, radius=1.5, colors = c)
cr = plt.Circle(xy=(0,0), radius=1, facecolor="white")
plt.gca().add_artist(cr) # to show
plt.show()
```

## Stem Plot

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,2,5,6,3,3]

plt.stem(x,y)

plt.show()
```
