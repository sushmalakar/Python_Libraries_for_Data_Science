## Bar Plot

```python
import numpy as np
```

```python
from matplotlib import pyplot as plt
```

```python
student = {"Sushmita Malakar" : 80, "Bishesh Lal Shrestha" : 95, "Sumit Malakar" : 82}
```

```python
print(student)
```

```python
type(student)
```

```python
names = list(student.keys())
values = list(student.values())
```

```python
plt.bar(names,values)
```

## Adding titles and labels to the bar plot

```python
plt.bar(names,values)
plt.title("Bar Plot")
plt.xlabel("Name of student")
plt.ylabel("Marks of student")
plt.show()
```

## Horizontal Bar Plot

```python
plt.barh(names,values, color="green")
plt.title("Bar Plot")
plt.xlabel("Name of student")
plt.ylabel("Marks of student")
plt.grid(True)
plt.show()
```

## Creating a Basic Scatter Plot

```python
import matplotlib.pyplot as plt
```

```python
x = [10,20,30,40,50,60,70,0,90]
a = [8,1,7,2,0,3,7,3,2]
```

```python
plt.scatter(x,a)
plt.show()
```

## Marker

```python
plt.scatter(x,a, marker="*", c="orange", s=50)
plt.show()
```

## Adding two markers in the same plot

```python
z = [10,20,30,40,50,60,70,0,90]
x = [8,1,7,2,0,3,7,3,2]
y = [7,2,5,6,9,1,4,5,3]
```

```python
plt.scatter(z,x, marker="*", c="g", s=50)
plt.scatter(z,y, marker=".", c="r", s=200)
plt.show()
```

## Adding two subplots

```python
z = [10,20,30,40,50,60,70,0,90]
x = [8,1,7,2,0,3,7,3,2]
y = [7,2,5,6,9,1,4,5,3]

plt.subplot(1,2,1)
plt.scatter(z,x,marker="*",c="g",s=100)

plt.subplot(1,2,2)
plt.scatter(z,y, marker=".", c="r", s=100)
plt.show()
```
