## Hexbin plot

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
tips = sns.load_dataset('tips')

# Set the aesthetic style
sns.set(style="ticks")

# Create Hexbin plot
plt.figure(figsize=(10,6))
plt.hexbin(tips['total_bill'], tips['tip'], gridsize=30, cmap='YlGnBu', edgecolors='black')
plt.colorbar(label='Number of Points in Bin')

# Add title and axis labels
plt.title("Hexbin Plot: Total Bill vs Tip", fontsize=16, fontweight='bold')
plt.xlabel("Total Bill ($)", fontsize=12)
plt.ylabel("Tip ($)", fontsize=12)

# Show the plot with better color contrast
plt.show()
```

## Quiver Plot

```python
import numpy as np
import matplotlib.pyplot as plt

# Set a high-quality style
plt.style.use('ggplot')

# Create a grid of points
x,y = np.meshgrid(np.linspace(-5,5,20), np.linspace(-5,5,20))

# Define the vector field for the fluid flow
u = -y # x-component of the vector (horizontal flow)
v = x #y-component of the vector (vertical flow)


# Calculate magnitude for color
magnitude = np.sqrt(u**2 + v**2)

# Create the Quiver Plot
plt.figure(figsize=(10,8))
quiver= plt.quiver(x,y,u,v,magnitude,angles='xy',scale_units='xy',scale=0.5,cmap='viridis',width=1.5)

# Enhance plot aesthetics
plt.title("Fluid Flow Vector Field",fontsize=18,fontweight='bold',color='darkgreen')
plt.xlabel("X-Axis",fontsize=14, color='darkred')
plt.ylabel("Y-Axis",fontsize=14, color='darkred')

# Add a colorbar with a label indicating magnitude
plt.colorbar(quiver, label='Magnitude', orientation ='vertical')

# Adjust grid and axis limts
plt.grid(True,linestyle='--',alpha=0.5)
plt.xlim([-5,5])
plt.ylim([-5,5])

# Show the plot
plt.show()
```

## Polar Plot 

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style
sns.set(style="white",palette="muted")

# Create data for the Polar plot
theta = np.linspace(0,2*np.pi,100)
r = np.abs(np.sin(2*theta)) * np.cos(2*theta)

# Create the Polar plot with improved aesthetics
plt.figure(figsize=(10,8))
ax = plt.subplot(111,projection='polar')

# Plot the curbe with enhanced color and styling
ax.plot(theta,r, color ='#1f77bd', linewidth=3,alpha=0.8,label='r =|sin(2θ) * cos(θ)')

# Fill the area under the curve with a soft transparent color
ax.fill(theta,r,color='#ff7f0e', alpha=0.3)

# Add title with larger font size and style
ax.set_title("Polar Plot: Sinusoidal Pattern", fontsize=20, fontweight='bold',color='darkblue',pad=20)

# Customize grid and ticks for better aesthetics
ax.grid(color='gray', linestyle='--',linewidth=0.6, alpha=0.5)
ax.set_xticks(np.linspace(0, 2*np.pi, 8)) # Custom angle ticks
ax.set_yticks([-1,0,1]) # Custom radial ticks for clarity

# Customize the labels with larger fonts and improved contrast
ax.set_xlabel("Angle (θ)",fontsize=14,color='darkgreen')
ax.set_ylabel("Magnitude (r)",fontsize=14,color='darkgreen')

# Add a legend to describe the plot
ax.legend(loc='upper right',fontsize=12)

# Remove the default ploar spine for a cleaner look
ax.spines['polar'].set_visible(False)

# Show the plot
plt.show()


```

## Heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Generate some sample data
data = np.random.rand(10,12) #10x12 matrix

# Setup the matplotlib figure
plt.figure(figsize=(12, 8))

# Create a heatmap with Seaborn
sns.heatmap(data,
            annot=True,       # Annotates each cell with its numeric values
            cmap='coolwarm',  # Choose a color palette
            center=0,         # Set the center of the colormap
            linewidths=0.5,   # Line width between cells
            linecolor='black', # Line color between cells
            cbar_kws={'shrink' : 0.75}, # colorbar size
            square=True, # Make the plot square-shaped
            fmt='.2f', # Number formatting in cells
            annot_kws={'size':10} # Font size for annotations
)

# Add lables and title
plt.title('Heatmap with Nice Aesthetics', fontsize=16)
plt.xlabel('Column', fontsize=12)
plt.ylabel('Row', fontsize=12)

# Show the plot
plt.tight_layout() # Adjust the layout to make sure everything fits
plt.show()
```

## Log-Log plot

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data
x = np.linspace(1,100,400)
y = x**2 

# Create a log-log plot
plt.figure(figsize=(6,4))
plt.loglog(x, y, label='$y = x^2$', color='b')

# Add labels and title
plt.xlabel('X', fontsize=12)
plt.ylabel('Y', fontsize=12)
plt.title('Log-Log Plot', fontsize=14)

# Show grid and legenf
plt.grid(True, which="both", linestyle="--", linewidth=0.50)
plt.legend()

# show plot
plt.tight_layout()
plt.show()
```

## Semi-log Plot

```python
x = np.linspace(1, 10, 100)
y = np.exp(x) # y = e^x, exponential growth

# Create a semi-log plot(logarithmic y-axis)
plt.figure(figsize=(6,4))
plt.semilogy(x, y, label='$y = e^x$', color = 'r')

# Add labels and title
plt.xlabel('X', fontsize=12)
plt.ylabel('Y (log scale)', fontsize=12)
plt.title('Semi-Log Plot (logarithmic Y)', fontsize=14)

# Show grid and legend
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend()

# Show plot
plt.tight_layout()
plt.show()
```
